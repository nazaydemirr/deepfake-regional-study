```python
# 3) Imports, reproducibility, run klasörleri ve yardımcı fonksiyonlar

import os
import sys
import json
import time
import random
import hashlib
import logging
import platform
import subprocess

from datetime import datetime
from pathlib import Path

import cv2
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image
from skimage.feature import hog, local_binary_pattern

from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    roc_auc_score,
    average_precision_score,
    roc_curve,
    precision_recall_curve,
    ConfusionMatrixDisplay,
)


# ---------------------------------------------------------
# Tekrar üretilebilirlik
# ---------------------------------------------------------

SEED = int(CONFIG["seed"])

random.seed(SEED)
np.random.seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)


# ---------------------------------------------------------
# Drive yolları
# ---------------------------------------------------------

DATA_ROOT = Path(CONFIG["data_root"])
ROI_METADATA_PATH = Path(CONFIG["roi_metadata_path"])
SELECTION_METADATA_PATH = Path(CONFIG["selection_metadata_path"])
RESULTS_ROOT = Path(CONFIG["results_root"])


# Girdi yolları doğrulanmadan sonuç klasörü oluşturulmaz.
assert DATA_ROOT.exists(), f"Ağız ROI klasörü bulunamadı: {DATA_ROOT}"
assert ROI_METADATA_PATH.exists(), (
    f"Ağız metadata dosyası bulunamadı: {ROI_METADATA_PATH}"
)
assert SELECTION_METADATA_PATH.exists(), (
    f"Seçim metadata dosyası bulunamadı: {SELECTION_METADATA_PATH}"
)


# ---------------------------------------------------------
# Deney kimliği ve sonuç klasörleri
# ---------------------------------------------------------

resume_run_id = CONFIG.get("resume_run_id")

if resume_run_id:
    RUN_ID = str(resume_run_id)
else:
    RUN_ID = (
        datetime.now().strftime("%Y%m%d_%H%M")
        + f"_mouth_hog_lbp_svm_seed{SEED}"
    )

RUN_DIR = RESULTS_ROOT / RUN_ID

if RUN_DIR.exists() and not resume_run_id:
    raise FileExistsError(
        f"Bu run_id zaten mevcut: {RUN_DIR}. "
        "Aynı dizine iki farklı deney yazılmaz. "
        "Yeni bir dakika bekleyin veya resume_run_id kullanın."
    )

DIRS = {
    "checkpoints": RUN_DIR / "checkpoints",
    "logs": RUN_DIR / "logs",
    "metrics": RUN_DIR / "metrics",
    "predictions": RUN_DIR / "predictions",
    "figures": RUN_DIR / "figures",
    "artifacts": RUN_DIR / "artifacts",
}

for path in [RUN_DIR, *DIRS.values()]:
    path.mkdir(parents=True, exist_ok=True)


# Kaynak ağız ROI klasörüne sonuç yazılmasını engelle.
assert RUN_DIR.resolve() != DATA_ROOT.resolve()
assert DATA_ROOT.resolve() not in RUN_DIR.resolve().parents


# ---------------------------------------------------------
# Logger
# ---------------------------------------------------------

LOG_PATH = DIRS["logs"] / "run.log"

logger = logging.getLogger("mouth_hog_lbp_svm")
logger.setLevel(logging.INFO)
logger.handlers.clear()

formatter = logging.Formatter(
    "%(asctime)s | %(levelname)s | %(message)s"
)

file_handler = logging.FileHandler(
    LOG_PATH,
    encoding="utf-8"
)
file_handler.setFormatter(formatter)
logger.addHandler(file_handler)

stream_handler = logging.StreamHandler(sys.stdout)
stream_handler.setFormatter(formatter)
logger.addHandler(stream_handler)


# ---------------------------------------------------------
# Güvenli JSON dönüşümü
# ---------------------------------------------------------

def json_safe(value):
    if isinstance(value, dict):
        return {
            str(key): json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [json_safe(item) for item in value]

    if isinstance(value, np.ndarray):
        return value.tolist()

    if isinstance(value, np.integer):
        return int(value)

    if isinstance(value, np.floating):
        return float(value)

    if isinstance(value, Path):
        return str(value)

    return value


# ---------------------------------------------------------
# Atomik dosya kayıt fonksiyonları
# ---------------------------------------------------------

def write_json_atomic(data, target: Path):
    target.parent.mkdir(parents=True, exist_ok=True)
    temporary = target.with_suffix(target.suffix + ".tmp")

    with temporary.open("w", encoding="utf-8") as file:
        json.dump(
            json_safe(data),
            file,
            ensure_ascii=False,
            indent=2
        )
        file.flush()
        os.fsync(file.fileno())

    with temporary.open("r", encoding="utf-8") as file:
        json.load(file)

    os.replace(temporary, target)


def write_csv_atomic(dataframe: pd.DataFrame, target: Path):
    target.parent.mkdir(parents=True, exist_ok=True)
    temporary = target.with_suffix(target.suffix + ".tmp")

    dataframe.to_csv(temporary, index=False)
    pd.read_csv(temporary)

    os.replace(temporary, target)


def save_npz_atomic(target: Path, **arrays):
    target.parent.mkdir(parents=True, exist_ok=True)
    temporary = target.with_suffix(target.suffix + ".tmp")

    with temporary.open("wb") as file:
        np.savez_compressed(file, **arrays)
        file.flush()
        os.fsync(file.fileno())

    with np.load(temporary, allow_pickle=False) as loaded:
        if not loaded.files:
            raise RuntimeError(
                f"NPZ doğrulaması başarısız: {temporary}"
            )

    os.replace(temporary, target)


def atomic_joblib_dump(obj, target: Path, required_keys=None):
    target.parent.mkdir(parents=True, exist_ok=True)
    temporary = target.with_suffix(target.suffix + ".tmp")

    joblib.dump(obj, temporary)
    loaded = joblib.load(temporary)

    if required_keys:
        valid = (
            isinstance(loaded, dict)
            and set(required_keys).issubset(loaded.keys())
        )

        if not valid:
            raise RuntimeError(
                f"Checkpoint doğrulaması başarısız: {temporary}"
            )

    os.replace(temporary, target)


def sha256_file(path: Path, chunk_size=1024 * 1024):
    digest = hashlib.sha256()

    with path.open("rb") as file:
        for chunk in iter(
            lambda: file.read(chunk_size),
            b""
        ):
            digest.update(chunk)

    return digest.hexdigest()


def save_figure(fig, target: Path):
    fig.tight_layout()

    fig.savefig(
        target,
        dpi=int(CONFIG["figure_dpi"]),
        bbox_inches="tight"
    )

    plt.close(fig)

    with Image.open(target) as image:
        width, height = image.size

    minimum_edge = int(
        CONFIG["minimum_figure_short_edge_px"]
    )

    assert min(width, height) >= minimum_edge, (
        f"Grafik çözünürlüğü yetersiz: {(width, height)}"
    )

    return width, height


# ---------------------------------------------------------
# Çözümlenmiş konfigürasyonu kaydet
# ---------------------------------------------------------

resolved_config = dict(CONFIG)

resolved_config.update({
    "run_id": RUN_ID,
    "run_dir": str(RUN_DIR),
    "created_at": datetime.now().isoformat(),
})

with (RUN_DIR / "config_resolved.yaml").open(
    "w",
    encoding="utf-8"
) as file:
    yaml.safe_dump(
        resolved_config,
        file,
        sort_keys=False,
        allow_unicode=True
    )


logger.info("RUN_ID=%s", RUN_ID)
logger.info("Ağız ROI girdisi=%s", DATA_ROOT)
logger.info("Deney çıktısı=%s", RUN_DIR)

print("\nBaşlangıç kontrolleri: PASS")
print("Run ID:", RUN_ID)
print("Sonuç klasörü:", RUN_DIR)
```

```python
# 4) Ağız metadata hazırlama, eski yolları düzeltme,
# veri muhasebesi ve kaynak-video sızıntısı kontrolleri

import re


# =========================================================
# 1. Metadata dosyalarını oku
# =========================================================

roi = pd.read_csv(ROI_METADATA_PATH)

selection = pd.read_csv(
    SELECTION_METADATA_PATH,
    encoding="utf-8-sig"
)


# =========================================================
# 2. Zorunlu sütunları doğrula
# =========================================================

required_roi_columns = {
    "sample_id",
    "source_frame",
    "relative_frame_path",
    "label",
    "split",
    "video_id",
    "frame_stem",
    "face_id",
    "faces_in_frame",
    "mouth_detected",
    "mouth_path",
    "debug_path",
    "landmarks_path",
    "mouth_bbox_x1",
    "mouth_bbox_y1",
    "mouth_bbox_x2",
    "mouth_bbox_y2",
    "mouth_width_px",
    "mouth_height_px",
    "landmark_count",
    "status",
    "error",
    "processing_ms",
}

required_selection_columns = {
    "sinif",
    "split",
    "orijinal_yol",
    "yeni_yol",
    "dosya_adi",
}

missing_roi_columns = (
    required_roi_columns - set(roi.columns)
)

missing_selection_columns = (
    required_selection_columns - set(selection.columns)
)

assert not missing_roi_columns, (
    "Ağız metadata dosyasında eksik sütunlar var: "
    f"{sorted(missing_roi_columns)}"
)

assert not missing_selection_columns, (
    "Seçim metadata dosyasında eksik sütunlar var: "
    f"{sorted(missing_selection_columns)}"
)


# =========================================================
# 3. Metadata içindeki eski mouth_path yollarını düzelt
# =========================================================
#
# Metadata yollarında eski üst klasör:
# /content/drive/MyDrive/AISC Çalışmaları/...
#
# Güncel üst klasör:
# /content/drive/MyDrive/AISC DeepFake Çalışmaları/...
#
# Drive'daki metadata.csv değiştirilmez.
# Yol yalnızca bu Colab oturumunda düzeltilir.

def resolve_mouth_path(raw_path):
    if pd.isna(raw_path):
        return ""

    raw_path = str(raw_path).strip()

    if raw_path == "":
        return ""

    path = Path(raw_path)
    path_parts = path.parts

    if "mouth_roi_output" in path_parts:
        root_index = path_parts.index(
            "mouth_roi_output"
        )

        relative_parts = path_parts[
            root_index + 1:
        ]

        resolved_path = DATA_ROOT.joinpath(
            *relative_parts
        )

        return str(resolved_path)

    return str(path)


roi = roi.copy()

roi["resolved_mouth_path"] = (
    roi["mouth_path"].apply(resolve_mouth_path)
)


# =========================================================
# 4. Kaynak frame ile seçim metadata'sını eşleştir
# =========================================================

selection = selection.copy()

selection["dosya_adi"] = (
    selection["dosya_adi"].astype(str)
)

assert selection["dosya_adi"].is_unique, (
    "secim_metadata.csv içinde dosya_adi "
    "değerleri benzersiz değil."
)


# Orijinal video adı gerçek kaynak yoldan alınır.
selection["source_video_resolved"] = (
    selection["orijinal_yol"].map(
        lambda path: Path(str(path)).parent.name
    )
)


def extract_original_frame_index(path):
    frame_stem = Path(str(path)).stem
    match = re.search(r"(\d+)$", frame_stem)

    if match is None:
        raise ValueError(
            "Frame numarası orijinal yoldan "
            f"çıkarılamadı: {path}"
        )

    return int(match.group(1))


selection["original_frame_index"] = (
    selection["orijinal_yol"].map(
        extract_original_frame_index
    )
)


# Ağız metadata'sındaki göreli frame yolundan
# yalnızca dosya adını al.
roi["selected_frame_name"] = (
    roi["relative_frame_path"].map(
        lambda path: Path(str(path)).name
    )
)


# Bir kaynak frame'den birden fazla yüz çıkabildiği için
# many_to_one eşleştirme kullanılır.
merged = roi.merge(
    selection[
        [
            "dosya_adi",
            "source_video_resolved",
            "original_frame_index",
            "orijinal_yol",
        ]
    ],
    left_on="selected_frame_name",
    right_on="dosya_adi",
    how="left",
    validate="many_to_one",
)


assert merged["source_video_resolved"].notna().all(), (
    "Bazı ağız ROI kayıtları "
    "secim_metadata.csv ile eşleşmedi."
)


# =========================================================
# 5. Ekip standardına uygun metadata oluştur
# =========================================================

status_values = (
    merged["status"]
    .fillna("ERROR")
    .astype(str)
    .str.upper()
)


metadata = pd.DataFrame({
    "sample_id": (
        merged["sample_id"].astype(str)
    ),

    "source_video": (
        merged["source_video_resolved"]
        .astype(str)
    ),

    "frame_index": (
        merged["original_frame_index"]
        .astype(int)
    ),

    "face_index": (
        merged["face_id"]
        .fillna(0)
        .astype(int)
    ),

    "roi_state": np.where(
        status_values == "SUCCESS",
        "mouth_detected",
        "mouth_not_detected",
    ),

    "label": (
        merged["label"]
        .astype(str)
        .str.lower()
    ),

    "split": (
        merged["split"]
        .astype(str)
        .str.lower()
    ),

    "status": status_values,

    "skip_reason": np.where(
        status_values == "SKIPPED",
        merged["error"]
        .fillna("MOUTH_NOT_DETECTED")
        .astype(str),
        "",
    ),

    # Başarılı ağız ROI dosyaları için aşağıda doldurulur.
    "sha256": "",

    "output_path": (
        merged["resolved_mouth_path"]
        .astype(str)
    ),

    # Model eğitim run kimliği
    "run_id": RUN_ID,

    "original_path": (
        merged["orijinal_yol"]
        .astype(str)
    ),

    "selected_frame_name": (
        merged["selected_frame_name"]
        .astype(str)
    ),

    "faces_in_frame": (
        merged["faces_in_frame"]
        .fillna(0)
        .astype(int)
    ),

    "mouth_width_px": (
        merged["mouth_width_px"]
    ),

    "mouth_height_px": (
        merged["mouth_height_px"]
    ),
})


# =========================================================
# 6. Metadata şemasını ve dağılımları doğrula
# =========================================================

assert len(metadata) == 3098, (
    "Beklenen metadata satır sayısı 3098, "
    f"bulunan: {len(metadata)}"
)

assert metadata["sample_id"].is_unique, (
    "Mükerrer sample_id tespit edildi."
)

assert set(metadata["label"].unique()) == {
    "real",
    "fake",
}, metadata["label"].value_counts().to_dict()

assert set(metadata["split"].unique()) == {
    "train",
    "val",
    "test",
}, metadata["split"].value_counts().to_dict()

assert set(
    metadata["status"].unique()
).issubset({
    "SUCCESS",
    "SKIPPED",
    "ERROR",
})


# =========================================================
# 7. Veri muhasebesi
# =========================================================

total_inputs = len(metadata)

success_count = int(
    (metadata["status"] == "SUCCESS").sum()
)

skipped_count = int(
    (metadata["status"] == "SKIPPED").sum()
)

error_count = int(
    (metadata["status"] == "ERROR").sum()
)


assert (
    total_inputs
    == success_count
    + skipped_count
    + error_count
), "Girdi-çıktı sayıları uyuşmuyor."


# Model yalnızca SUCCESS durumundaki ağız ROI'lerini kullanır.
eligible = metadata.loc[
    metadata["status"] == "SUCCESS"
].copy().reset_index(drop=True)


assert len(eligible) == 2987, (
    "Beklenen başarılı ağız ROI sayısı 2987, "
    f"bulunan: {len(eligible)}"
)


# =========================================================
# 8. Başarılı ağız ROI dosyalarının varlığını doğrula
# =========================================================

missing_success_files = [
    path
    for path in eligible["output_path"]
    if not Path(path).exists()
]


assert not missing_success_files, (
    "SUCCESS durumundaki "
    f"{len(missing_success_files)} ağız ROI "
    "dosyası diskte bulunamadı."
)


# =========================================================
# 9. Dosya bütünlük hash'lerini hesapla
# =========================================================
#
# DÜZELTME:
# sha256_file fonksiyonuna metin yerine Path nesnesi verilir.

eligible["sha256"] = (
    eligible["output_path"].map(
        lambda path: sha256_file(Path(path))
    )
)


# Hesaplanan hash değerlerini ana metadata tablosuna aktar.
success_mask = (
    metadata["status"] == "SUCCESS"
)

metadata.loc[
    success_mask,
    "sha256",
] = eligible["sha256"].to_numpy()


assert (
    eligible["sha256"].astype(str).str.len() == 64
).all(), "Geçersiz SHA-256 değeri tespit edildi."


# =========================================================
# 10. Kaynak-video düzeyinde veri sızıntısı kontrolü
# =================================================================

video_sets = {
    split_name: set(
        eligible.loc[
            eligible["split"] == split_name,
            "source_video",
        ].astype(str)
    )
    for split_name in [
        "train",
        "val",
        "test",
    ]
}


intersections = {
    "train_val": len(
        video_sets["train"]
        & video_sets["val"]
    ),

    "train_test": len(
        video_sets["train"]
        & video_sets["test"]
    ),

    "val_test": len(
        video_sets["val"]
        & video_sets["test"]
    ),
}


assert intersections["train_val"] == 0, (
    "Train ve validation arasında "
    "kaynak-video sızıntısı var."
)

assert intersections["train_test"] == 0, (
    "Train ve test arasında "
    "kaynak-video sızıntısı var."
)

assert intersections["val_test"] == 0, (
    "Validation ve test arasında "
    "kaynak-video sızıntısı var."
)


# Bir kaynak video yalnızca tek splitte bulunmalıdır.
assert (
    eligible.groupby("source_video")["split"]
    .nunique()
    .max()
    == 1
), "Aynı kaynak video birden fazla splitte bulunuyor."


# Bir kaynak video yalnızca tek gerçek sınıfa ait olmalıdır.
assert (
    eligible.groupby("source_video")["label"]
    .nunique()
    .max()
    == 1
), "Aynı kaynak video birden fazla label ile eşleşiyor."


# =========================================================
# 11. Split ve sınıf dağılımları
# =========================================================

split_class_counts = (
    eligible.groupby(["split", "label"])
    .size()
    .unstack(fill_value=0)
    .to_dict(orient="index")
)


split_source_video_counts = {
    split_name: int(
        eligible.loc[
            eligible["split"] == split_name,
            "source_video",
        ].nunique()
    )
    for split_name in [
        "train",
        "val",
        "test",
    ]
}


# =========================================================
# 12. Veri muhasebesi sonuçlarını kaydet
# =========================================================

accounting = {
    "run_id": RUN_ID,

    "total_metadata_rows": total_inputs,

    "unique_source_frames": int(
        metadata["selected_frame_name"].nunique()
    ),

    "success_count": success_count,

    "skipped_count": skipped_count,

    "error_count": error_count,

    "training_eligible_success_count": int(
        len(eligible)
    ),

    "unique_source_videos": int(
        eligible["source_video"].nunique()
    ),

    "split_class_counts": split_class_counts,

    "split_source_video_counts": (
        split_source_video_counts
    ),

    "source_video_intersections": (
        intersections
    ),
}


write_json_atomic(
    accounting,
    DIRS["metrics"] / "data_accounting.json"
)


write_csv_atomic(
    metadata,
    DIRS["artifacts"] / "metadata_used.csv"
)


# =========================================================
# 13. Sonuçları göster
# =========================================================

print(json.dumps(
    accounting,
    ensure_ascii=False,
    indent=2
))

print(
    "\nMetadata and leakage quality gates: PASS"
)
```

```python
# 5) HOG + LBP özellik çıkarımı ve klasik ML smoke testi

IMG_W = int(CONFIG["image_width"])
IMG_H = int(CONFIG["image_height"])

HOG_CFG = CONFIG["hog"]
LBP_CFG = CONFIG["lbp"]


# =========================================================
# 1. Ağız ROI görüntüsünü oku ve ön işle
# =========================================================

def load_preprocessed_grayscale(path: str) -> np.ndarray:
    image = cv2.imread(
        str(path),
        cv2.IMREAD_GRAYSCALE
    )

    if image is None:
        raise ValueError(
            f"Ağız ROI görüntüsü okunamadı: {path}"
        )

    image = cv2.resize(
        image,
        (IMG_W, IMG_H),
        interpolation=cv2.INTER_AREA
    )

    if image.dtype != np.uint8:
        image = np.clip(
            image,
            0,
            255
        ).astype(np.uint8)

    return image


# =========================================================
# 2. HOG özelliklerini çıkar
# =========================================================

def extract_hog(gray: np.ndarray) -> np.ndarray:
    features = hog(
        gray,
        orientations=int(
            HOG_CFG["orientations"]
        ),
        pixels_per_cell=tuple(
            HOG_CFG["pixels_per_cell"]
        ),
        cells_per_block=tuple(
            HOG_CFG["cells_per_block"]
        ),
        block_norm=str(
            HOG_CFG["block_norm"]
        ),
        feature_vector=True,
    )

    return np.asarray(
        features,
        dtype=np.float32
    )


# =========================================================
# 3. LBP özelliklerini çıkar
# =========================================================

def extract_lbp(gray: np.ndarray) -> np.ndarray:
    points = int(LBP_CFG["points"])
    radius = int(LBP_CFG["radius"])
    method = str(LBP_CFG["method"])

    lbp_image = local_binary_pattern(
        gray,
        P=points,
        R=radius,
        method=method
    )

    if method == "uniform":
        number_of_bins = points + 2
    else:
        number_of_bins = int(
            lbp_image.max() + 1
        )

    histogram, _ = np.histogram(
        lbp_image.ravel(),
        bins=np.arange(
            0,
            number_of_bins + 1
        ),
        range=(0, number_of_bins)
    )

    histogram = histogram.astype(
        np.float32
    )

    histogram /= (
        histogram.sum() + 1e-12
    )

    return histogram


# =========================================================
# 4. HOG ve LBP özelliklerini birleştir
# =========================================================

def extract_features(path: str):
    gray = load_preprocessed_grayscale(
        path
    )

    hog_features = extract_hog(gray)
    lbp_features = extract_lbp(gray)

    fused_features = np.concatenate(
        [
            hog_features,
            lbp_features,
        ]
    ).astype(np.float32)

    if not np.isfinite(
        fused_features
    ).all():
        raise FloatingPointError(
            "HOG/LBP özelliklerinde NaN veya Inf "
            f"tespit edildi: {path}"
        )

    return (
        hog_features,
        lbp_features,
        fused_features,
    )


# =========================================================
# 5. Smoke test için iki Real ve iki Fake train örneği
# =========================================================

smoke_rows = pd.concat(
    [
        eligible[
            (eligible["split"] == "train")
            & (eligible["label"] == "real")
        ].head(2),

        eligible[
            (eligible["split"] == "train")
            & (eligible["label"] == "fake")
        ].head(2),
    ],
    ignore_index=True
)


assert len(smoke_rows) == 4, (
    "Smoke test için dört örnek bulunamadı."
)

assert smoke_rows["label"].nunique() == 2, (
    "Smoke test iki sınıfı da içermiyor."
)


# =========================================================
# 6. Özellik çıkarımı + SVM fit/predict smoke testi
# =========================================================

smoke_X = []
smoke_y = []

for row in smoke_rows.itertuples(
    index=False
):
    _, _, fused = extract_features(
        row.output_path
    )

    smoke_X.append(fused)

    smoke_y.append(
        1
        if row.label
        == CONFIG["positive_class"]
        else 0
    )


smoke_X = np.vstack(smoke_X)

smoke_y = np.asarray(
    smoke_y,
    dtype=np.int64
)


assert np.isfinite(smoke_X).all(), (
    "Smoke test özelliklerinde NaN/Inf var."
)


smoke_scaler = StandardScaler()

smoke_X_scaled = smoke_scaler.fit_transform(
    smoke_X
)


smoke_model = SVC(
    kernel="linear",
    C=1.0,
    class_weight="balanced"
)

smoke_model.fit(
    smoke_X_scaled,
    smoke_y
)


smoke_predictions = smoke_model.predict(
    smoke_X_scaled
)


assert len(smoke_predictions) == len(
    smoke_y
)


print("Classical-ML smoke test: PASS")
print("Smoke sample count:", len(smoke_y))
print("HOG dimension:", len(extract_hog(
    load_preprocessed_grayscale(
        smoke_rows.iloc[0]["output_path"]
    )
)))
print("LBP dimension:", len(extract_lbp(
    load_preprocessed_grayscale(
        smoke_rows.iloc[0]["output_path"]
    )
)))
print("Fused feature dimension:", smoke_X.shape[1])
```

```python
# 6) Train, validation ve test için HOG + LBP özellik çıkarımı
# Her split tamamlandığında atomik NPZ cache olarak Drive'a kaydedilir.

def build_split_features(split_name: str):
    cache_path = (
        DIRS["artifacts"]
        / f"features_{split_name}.npz"
    )

    # Split daha önce tamamlandıysa tekrar hesaplama.
    if cache_path.exists():
        logger.info(
            "Cached features loaded: %s",
            cache_path
        )

        with np.load(
            cache_path,
            allow_pickle=False
        ) as loaded:
            return {
                "X_hog": loaded["X_hog"],
                "X_lbp": loaded["X_lbp"],
                "X_fused": loaded["X_fused"],
                "y": loaded["y"],
                "sample_id": (
                    loaded["sample_id"].astype(str)
                ),
                "source_video": (
                    loaded["source_video"].astype(str)
                ),
                "frame_index": (
                    loaded["frame_index"]
                ),
                "output_path": (
                    loaded["output_path"].astype(str)
                ),
            }

    split_data = eligible.loc[
        eligible["split"] == split_name
    ].copy().reset_index(drop=True)

    assert len(split_data) > 0, (
        f"{split_name} splitinde örnek bulunamadı."
    )

    hog_rows = []
    lbp_rows = []
    fused_rows = []
    labels = []

    start_time = time.time()

    for index, row in enumerate(
        split_data.itertuples(index=False),
        start=1
    ):
        hog_features, lbp_features, fused_features = (
            extract_features(row.output_path)
        )

        hog_rows.append(hog_features)
        lbp_rows.append(lbp_features)
        fused_rows.append(fused_features)

        labels.append(
            1
            if row.label
            == CONFIG["positive_class"]
            else 0
        )

        if (
            index % 100 == 0
            or index == len(split_data)
        ):
            elapsed_minutes = (
                time.time() - start_time
            ) / 60

            logger.info(
                "Feature extraction %s: %d/%d | %.2f min",
                split_name,
                index,
                len(split_data),
                elapsed_minutes,
            )

    split_features = {
        "X_hog": np.vstack(
            hog_rows
        ).astype(np.float32),

        "X_lbp": np.vstack(
            lbp_rows
        ).astype(np.float32),

        "X_fused": np.vstack(
            fused_rows
        ).astype(np.float32),

        "y": np.asarray(
            labels,
            dtype=np.int64
        ),

        "sample_id": split_data[
            "sample_id"
        ].astype(str).to_numpy(),

        "source_video": split_data[
            "source_video"
        ].astype(str).to_numpy(),

        "frame_index": split_data[
            "frame_index"
        ].to_numpy(dtype=np.int64),

        "output_path": split_data[
            "output_path"
        ].astype(str).to_numpy(),
    }

    assert np.isfinite(
        split_features["X_fused"]
    ).all(), (
        f"{split_name} özelliklerinde NaN/Inf var."
    )

    assert (
        len(split_features["X_fused"])
        == len(split_data)
    )

    save_npz_atomic(
        cache_path,
        **split_features
    )

    logger.info(
        "Feature extraction %s completed in %.2f min",
        split_name,
        (time.time() - start_time) / 60,
    )

    return split_features


features = {}

for split_name in [
    "train",
    "val",
    "test",
]:
    features[split_name] = (
        build_split_features(split_name)
    )


# Temel boyut doğrulamaları
feature_dimensions = {
    "hog": int(
        features["train"]["X_hog"].shape[1]
    ),
    "lbp": int(
        features["train"]["X_lbp"].shape[1]
    ),
    "fused": int(
        features["train"]["X_fused"].shape[1]
    ),
}


assert feature_dimensions == {
    "hog": 3780,
    "lbp": 10,
    "fused": 3790,
}, feature_dimensions


for split_name in [
    "train",
    "val",
    "test",
]:
    print(
        split_name,
        features[split_name]["X_fused"].shape,
        "Real:",
        int(
            (
                features[split_name]["y"] == 0
            ).sum()
        ),
        "Fake:",
        int(
            (
                features[split_name]["y"] == 1
            ).sum()
        ),
    )

print("\nFeature extraction quality gate: PASS")
print("Feature dimensions:", feature_dimensions)
```

```python
# 7) Train-only StandardScaler +
# validation-only SVM ve threshold seçimi

# =========================================================
# 1. Train, validation ve test özelliklerini hazırla
# =========================================================

X_train = features["train"]["X_fused"]
y_train = features["train"]["y"]

X_val = features["val"]["X_fused"]
y_val = features["val"]["y"]

X_test = features["test"]["X_fused"]
y_test = features["test"]["y"]


assert X_train.shape[1] == 3790
assert X_val.shape[1] == 3790
assert X_test.shape[1] == 3790

assert set(np.unique(y_train)) == {0, 1}
assert set(np.unique(y_val)) == {0, 1}
assert set(np.unique(y_test)) == {0, 1}


# =========================================================
# 2. StandardScaler yalnızca train üzerinde öğrenilir
# =========================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(
    X_train
)

X_val_scaled = scaler.transform(
    X_val
)


# Test burada ölçeklenmez.
# Test setine yalnızca nihai değerlendirme hücresinde bakılacak.

assert np.isfinite(
    X_train_scaled
).all()

assert np.isfinite(
    X_val_scaled
).all()


feature_pipeline = {
    "input": "mouth ROI",
    "grayscale": True,
    "resize": [
        int(CONFIG["image_width"]),
        int(CONFIG["image_height"]),
    ],
    "hog": CONFIG["hog"],
    "lbp": CONFIG["lbp"],
    "fusion": "HOG + LBP concatenation",
    "scaler": "StandardScaler fitted on train only",
    "feature_dimensions": feature_dimensions,
}


# =========================================================
# 3. Değerlendirme metrikleri
# =========================================================

def metrics_from_scores(
    y_true,
    decision_scores,
    threshold
):
    y_true = np.asarray(y_true)
    decision_scores = np.asarray(
        decision_scores
    )

    y_prediction = (
        decision_scores >= threshold
    ).astype(np.int64)

    tn, fp, fn, tp = confusion_matrix(
        y_true,
        y_prediction,
        labels=[0, 1]
    ).ravel()

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0.0
    )

    recall = recall_score(
        y_true,
        y_prediction,
        zero_division=0
    )

    balanced_accuracy = (
        recall + specificity
    ) / 2.0

    return {
        "accuracy": accuracy_score(
            y_true,
            y_prediction
        ),

        "precision": precision_score(
            y_true,
            y_prediction,
            zero_division=0
        ),

        "recall": recall,

        "specificity": specificity,

        "balanced_accuracy": (
            balanced_accuracy
        ),

        "f1": f1_score(
            y_true,
            y_prediction,
            zero_division=0
        ),

        "roc_auc": roc_auc_score(
            y_true,
            decision_scores
        ),

        "average_precision": (
            average_precision_score(
                y_true,
                decision_scores
            )
        ),

        "tn": int(tn),
        "fp": int(fp),
        "fn": int(fn),
        "tp": int(tp),
    }


# =========================================================
# 4. Validation setinde en iyi F1 threshold'unu bul
# =========================================================

def find_best_f1_threshold(
    y_true,
    decision_scores
):
    precision_values, recall_values, thresholds = (
        precision_recall_curve(
            y_true,
            decision_scores
        )
    )

    if len(thresholds) == 0:
        return 0.0

    f1_values = (
        2
        * precision_values[:-1]
        * recall_values[:-1]
        / (
            precision_values[:-1]
            + recall_values[:-1]
            + 1e-12
        )
    )

    best_index = int(
        np.nanargmax(f1_values)
    )

    return float(
        thresholds[best_index]
    )


# =========================================================
# 5. SVM adaylarını oluştur
# =========================================================

svm_config = CONFIG["svm"]

candidates = []

for kernel in svm_config["kernels"]:
    for c_value in svm_config["c_values"]:

        if kernel == "linear":
            candidates.append({
                "kernel": "linear",
                "C": float(c_value),
                "gamma": None,
            })

        elif kernel == "rbf":
            for gamma_value in (
                svm_config["gamma_values"]
            ):
                candidates.append({
                    "kernel": "rbf",
                    "C": float(c_value),
                    "gamma": gamma_value,
                })

        else:
            raise ValueError(
                f"Desteklenmeyen kernel: {kernel}"
            )


assert len(candidates) == 12, (
    f"Beklenen SVM aday sayısı 12, "
    f"bulunan: {len(candidates)}"
)

print("SVM candidate count:", len(candidates))


# =========================================================
# 6. Bütün adayları yalnızca validation ile değerlendir
# =========================================================

search_rows = []
search_start_time = time.time()

for candidate_index, parameters in enumerate(
    candidates,
    start=1
):
    model_parameters = {
        "kernel": parameters["kernel"],
        "C": parameters["C"],
        "class_weight": (
            svm_config["class_weight"]
        ),
        "probability": False,
        "cache_size": 2048,
    }

    if parameters["kernel"] == "rbf":
        model_parameters["gamma"] = (
            parameters["gamma"]
        )

    candidate_model = SVC(
        **model_parameters
    )

    candidate_start_time = time.time()

    candidate_model.fit(
        X_train_scaled,
        y_train
    )

    validation_scores = (
        candidate_model.decision_function(
            X_val_scaled
        )
    )

    assert np.isfinite(
        validation_scores
    ).all(), (
        "Validation karar skorlarında "
        "NaN veya Inf tespit edildi."
    )

    candidate_threshold = (
        find_best_f1_threshold(
            y_val,
            validation_scores
        )
    )

    candidate_metrics = (
        metrics_from_scores(
            y_val,
            validation_scores,
            candidate_threshold
        )
    )

    search_row = {
        **parameters,
        "threshold": candidate_threshold,

        **{
            f"val_{metric_name}": float(
                metric_value
            )
            for metric_name, metric_value
            in candidate_metrics.items()
        },

        "fit_seconds": (
            time.time()
            - candidate_start_time
        ),
    }

    search_rows.append(search_row)

    # Her adaydan sonra sonuçları Drive'a kaydet.
    write_csv_atomic(
        pd.DataFrame(search_rows),
        DIRS["metrics"]
        / "svm_validation_search.csv"
    )

    logger.info(
        "SVM candidate %d/%d | %s | val_f1=%.4f",
        candidate_index,
        len(candidates),
        parameters,
        candidate_metrics["f1"],
    )


# =========================================================
# 7. En iyi SVM adayını validation sonucuna göre seç
# =========================================================

search_dataframe = pd.DataFrame(
    search_rows
)

search_dataframe = (
    search_dataframe.sort_values(
        [
            "val_f1",
            "val_roc_auc",
            "val_accuracy",
        ],
        ascending=[
            False,
            False,
            False,
        ],
    )
    .reset_index(drop=True)
)


best_row = (
    search_dataframe.iloc[0].to_dict()
)


best_model_parameters = {
    "kernel": best_row["kernel"],
    "C": float(best_row["C"]),
    "class_weight": (
        svm_config["class_weight"]
    ),
    "probability": False,
    "cache_size": 2048,
}


if best_row["kernel"] == "rbf":
    best_model_parameters["gamma"] = (
        best_row["gamma"]
    )


# En iyi parametrelerle model yalnızca train üzerinde eğitilir.
best_model = SVC(
    **best_model_parameters
)

best_model.fit(
    X_train_scaled,
    y_train
)


selected_threshold = float(
    best_row["threshold"]
)


# =========================================================
# 8. Model seçim özetini oluştur
# =========================================================

raw_gamma = best_row.get("gamma")

if pd.isna(raw_gamma):
    selected_gamma = None
else:
    selected_gamma = raw_gamma


selection_summary = {
    "selection_set": "validation",

    "selection_metric": (
        svm_config["selection_metric"]
    ),

    "kernel": best_row["kernel"],

    "C": float(best_row["C"]),

    "gamma": selected_gamma,

    "threshold": selected_threshold,

    "validation_metrics": {
        key.replace("val_", ""): float(value)
        for key, value in best_row.items()
        if str(key).startswith("val_")
    },

    "candidate_count": len(candidates),

    "search_seconds": (
        time.time() - search_start_time
    ),

    "test_set_used_for_selection": False,
}


write_json_atomic(
    selection_summary,
    DIRS["metrics"] / "svm_selection.json"
)


# =========================================================
# 9. Kullanılabilir model dosyasını atomik kaydet
# =========================================================

model_artifact = {
    "model": best_model,
    "scaler": scaler,
    "threshold": selected_threshold,
    "positive_class": (
        CONFIG["positive_class"]
    ),
    "feature_pipeline": feature_pipeline,
    "config": resolved_config,
}


atomic_joblib_dump(
    model_artifact,
    DIRS["artifacts"] / "svm_model.joblib",
    required_keys=[
        "model",
        "scaler",
        "threshold",
    ],
)


# =========================================================
# 10. Sonucu göster
# =========================================================

print("\nSelected SVM:")
print(json.dumps(
    json_safe(selection_summary),
    ensure_ascii=False,
    indent=2
))

print(
    "\nTrain-only scaling and "
    "validation-only model selection: PASS"
)
```

```python
# 9) Nihai test değerlendirmesi
# Test seti model veya threshold seçiminde kullanılmaz.

# =========================================================
# 1. Örnek bazlı tahmin tablosu oluştur
# =========================================================

def build_prediction_dataframe(
    split_name,
    split_features,
    decision_scores,
    threshold
):
    true_labels = split_features["y"]

    predicted_labels = (
        decision_scores >= threshold
    ).astype(np.int64)

    return pd.DataFrame({
        "sample_id": (
            split_features["sample_id"]
        ),

        "source_video": (
            split_features["source_video"]
        ),

        "frame_index": (
            split_features["frame_index"]
        ),

        "split": split_name,

        "y_true": true_labels,

        "label": np.where(
            true_labels == 1,
            CONFIG["positive_class"],
            "real",
        ),

        "decision_score": decision_scores,

        "threshold": threshold,

        "y_pred": predicted_labels,

        "predicted_label": np.where(
            predicted_labels == 1,
            CONFIG["positive_class"],
            "real",
        ),

        "correct": (
            true_labels == predicted_labels
        ),

        "output_path": (
            split_features["output_path"]
        ),

        "run_id": RUN_ID,
    })


# =========================================================
# 2. Frame tahminlerini kaynak-video düzeyinde birleştir
# =========================================================

def aggregate_video_level(
    frame_dataframe: pd.DataFrame
):
    label_counts = (
        frame_dataframe.groupby(
            "source_video"
        )["y_true"].nunique()
    )

    assert label_counts.max() == 1, (
        "Bir kaynak videoda çelişkili "
        "gerçek sınıflar bulundu."
    )

    video_dataframe = (
        frame_dataframe.groupby(
            "source_video",
            as_index=False
        ).agg(
            y_true=("y_true", "first"),

            decision_score=(
                "decision_score",
                "mean"
            ),

            frame_count=(
                "sample_id",
                "size"
            ),
        )
    )

    video_dataframe["threshold"] = (
        selected_threshold
    )

    video_dataframe["y_pred"] = (
        video_dataframe["decision_score"]
        >= selected_threshold
    ).astype(np.int64)

    video_dataframe["label"] = np.where(
        video_dataframe["y_true"] == 1,
        CONFIG["positive_class"],
        "real",
    )

    video_dataframe["predicted_label"] = (
        np.where(
            video_dataframe["y_pred"] == 1,
            CONFIG["positive_class"],
            "real",
        )
    )

    video_dataframe["correct"] = (
        video_dataframe["y_true"]
        == video_dataframe["y_pred"]
    )

    video_dataframe["run_id"] = RUN_ID

    return video_dataframe


# =========================================================
# 3. Test özelliklerini train scaler ile dönüştür
# =========================================================

X_test_scaled = scaler.transform(
    X_test
)

assert np.isfinite(
    X_test_scaled
).all()


# =========================================================
# 4. Nihai test skorlarını üret
# =========================================================

test_scores = best_model.decision_function(
    X_test_scaled
)

assert np.isfinite(test_scores).all()


# =========================================================
# 5. Frame-level ve video-level tahminleri oluştur
# =========================================================

frame_predictions = (
    build_prediction_dataframe(
        "test",
        features["test"],
        test_scores,
        selected_threshold,
    )
)


video_predictions = (
    aggregate_video_level(
        frame_predictions
    )
)


# =========================================================
# 6. Nihai test metriklerini hesapla
# =========================================================

frame_metrics = metrics_from_scores(
    frame_predictions[
        "y_true"
    ].to_numpy(),

    frame_predictions[
        "decision_score"
    ].to_numpy(),

    selected_threshold,
)


video_metrics = metrics_from_scores(
    video_predictions[
        "y_true"
    ].to_numpy(),

    video_predictions[
        "decision_score"
    ].to_numpy(),

    selected_threshold,
)


# =========================================================
# 7. Tahmin ve metrik dosyalarını kaydet
# =========================================================

write_csv_atomic(
    frame_predictions,
    DIRS["predictions"]
    / "test_predictions_frame_level.csv"
)


write_csv_atomic(
    video_predictions,
    DIRS["predictions"]
    / "test_predictions_video_level.csv"
)


write_json_atomic(
    frame_metrics,
    DIRS["metrics"]
    / "test_metrics_frame_level.json"
)


write_json_atomic(
    video_metrics,
    DIRS["metrics"]
    / "test_metrics_video_level.json"
)


metrics_summary = pd.DataFrame([
    {
        "level": "frame",
        **frame_metrics,
    },
    {
        "level": "video",
        **video_metrics,
    },
])


write_csv_atomic(
    metrics_summary,
    DIRS["metrics"]
    / "test_metrics_summary.csv"
)


# =========================================================
# 8. Sonuçları göster
# =========================================================

print("FINAL TEST METRICS")
print(
    metrics_summary.to_string(
        index=False
    )
)

print(
    "\nFrame prediction count:",
    len(frame_predictions)
)

print(
    "Video prediction count:",
    len(video_predictions)
)

print(
    "\nFrozen test evaluation: PASS"
)
```

```python
# 10) İngilizce ve yüksek çözünürlüklü deney grafikleri

figure_records = []


# =========================================================
# 1. Grafik kayıt ve kalite kontrol fonksiyonu
# =========================================================

def register_figure(
    figure,
    filename
):
    figure_path = (
        DIRS["figures"] / filename
    )

    width_px, height_px = save_figure(
        figure,
        figure_path
    )

    figure_records.append({
        "file": filename,
        "width_px": width_px,
        "height_px": height_px,
        "short_edge_px": min(
            width_px,
            height_px
        ),
        "quality_pass": (
            min(width_px, height_px)
            >= int(
                CONFIG[
                    "minimum_figure_short_edge_px"
                ]
            )
        ),
    })


# Ortak renkler
REAL_COLOR = "#3182CE"
FAKE_COLOR = "#DD6B20"
MODEL_COLOR = "#2F855A"
REFERENCE_COLOR = "#718096"


# =========================================================
# 2. Dataset dağılım grafiği
# =========================================================

count_dataframe = (
    eligible.groupby(
        ["split", "label"]
    )
    .size()
    .unstack(fill_value=0)
    .reindex([
        "train",
        "val",
        "test",
    ])
)


figure, axis = plt.subplots(
    figsize=(10, 6),
    dpi=int(CONFIG["figure_dpi"])
)

x_positions = np.arange(
    len(count_dataframe.index)
)

bar_width = 0.35

axis.bar(
    x_positions - bar_width / 2,
    count_dataframe.get(
        "real",
        pd.Series(
            0,
            index=count_dataframe.index
        )
    ),
    bar_width,
    label="Real",
    color=REAL_COLOR,
)

axis.bar(
    x_positions + bar_width / 2,
    count_dataframe.get(
        "fake",
        pd.Series(
            0,
            index=count_dataframe.index
        )
    ),
    bar_width,
    label="Fake",
    color=FAKE_COLOR,
)

axis.set_title(
    "Mouth ROI Dataset Distribution by Split and Class",
    fontsize=14,
    fontweight="bold",
    pad=12,
)

axis.set_xlabel(
    "Dataset Split",
    fontsize=11
)

axis.set_ylabel(
    "Number of Mouth ROI Samples",
    fontsize=11
)

axis.set_xticks(
    x_positions,
    [
        "Train",
        "Validation",
        "Test",
    ]
)

axis.legend(
    frameon=True
)

axis.grid(
    True,
    axis="y",
    alpha=0.25
)

register_figure(
    figure,
    "dataset_distribution.png"
)


# =========================================================
# 3. Validation SVM aday karşılaştırması
# =========================================================

plot_search = (
    search_dataframe.copy()
    .sort_values(
        "val_f1",
        ascending=False
    )
    .reset_index(drop=True)
)


candidate_labels = []

for row in plot_search.itertuples(
    index=False
):
    label_text = (
        f"{row.kernel} | C={row.C}"
    )

    if row.kernel == "rbf":
        label_text += (
            f" | gamma={row.gamma}"
        )

    candidate_labels.append(
        label_text
    )


figure, axis = plt.subplots(
    figsize=(12, 7),
    dpi=int(CONFIG["figure_dpi"])
)

axis.bar(
    np.arange(len(plot_search)),
    plot_search["val_f1"],
    color=MODEL_COLOR,
)

axis.set_title(
    "Validation F1 Score by SVM Candidate",
    fontsize=14,
    fontweight="bold",
    pad=12,
)

axis.set_xlabel(
    "SVM Hyperparameter Candidate",
    fontsize=11
)

axis.set_ylabel(
    "Validation F1 Score",
    fontsize=11
)

axis.set_xticks(
    np.arange(len(candidate_labels)),
    candidate_labels,
    rotation=60,
    ha="right",
)

axis.grid(
    True,
    axis="y",
    alpha=0.25
)

register_figure(
    figure,
    "svm_validation_f1_comparison.png"
)


# =========================================================
# 4. Confusion matrix fonksiyonu
# =========================================================

def create_confusion_matrix_plot(
    prediction_dataframe,
    level_name,
    filename
):
    matrix = confusion_matrix(
        prediction_dataframe["y_true"],
        prediction_dataframe["y_pred"],
        labels=[0, 1],
    )

    figure, axis = plt.subplots(
        figsize=(8, 8),
        dpi=int(CONFIG["figure_dpi"])
    )

    display = ConfusionMatrixDisplay(
        matrix,
        display_labels=[
            "Real",
            "Fake",
        ],
    )

    display.plot(
        ax=axis,
        colorbar=False,
        values_format="d",
        cmap="Blues",
    )

    axis.set_title(
        f"Test Confusion Matrix - {level_name}",
        fontsize=14,
        fontweight="bold",
        pad=12,
    )

    register_figure(
        figure,
        filename
    )


# =========================================================
# 5. ROC curve fonksiyonu
# =========================================================

def create_roc_plot(
    prediction_dataframe,
    level_name,
    filename
):
    false_positive_rate, true_positive_rate, _ = (
        roc_curve(
            prediction_dataframe["y_true"],
            prediction_dataframe[
                "decision_score"
            ],
        )
    )

    auc_value = roc_auc_score(
        prediction_dataframe["y_true"],
        prediction_dataframe[
            "decision_score"
        ],
    )

    figure, axis = plt.subplots(
        figsize=(10, 6),
        dpi=int(CONFIG["figure_dpi"])
    )

    axis.plot(
        false_positive_rate,
        true_positive_rate,
        linewidth=2,
        color=MODEL_COLOR,
        label=f"ROC AUC = {auc_value:.3f}",
    )

    axis.plot(
        [0, 1],
        [0, 1],
        linestyle="--",
        linewidth=1.5,
        color=REFERENCE_COLOR,
        label="Random Classifier",
    )

    axis.set_title(
        f"Test ROC Curve - {level_name}",
        fontsize=14,
        fontweight="bold",
        pad=12,
    )

    axis.set_xlabel(
        "False Positive Rate",
        fontsize=11
    )

    axis.set_ylabel(
        "True Positive Rate",
        fontsize=11
    )

    axis.legend(
        frameon=True,
        loc="lower right"
    )

    axis.grid(
        True,
        alpha=0.25
    )

    register_figure(
        figure,
        filename
    )


# =========================================================
# 6. Precision-recall curve fonksiyonu
# =========================================================

def create_precision_recall_plot(
    prediction_dataframe,
    level_name,
    filename
):
    precision_values, recall_values, _ = (
        precision_recall_curve(
            prediction_dataframe["y_true"],
            prediction_dataframe[
                "decision_score"
            ],
        )
    )

    average_precision = (
        average_precision_score(
            prediction_dataframe["y_true"],
            prediction_dataframe[
                "decision_score"
            ],
        )
    )

    figure, axis = plt.subplots(
        figsize=(10, 6),
        dpi=int(CONFIG["figure_dpi"])
    )

    axis.plot(
        recall_values,
        precision_values,
        linewidth=2,
        color=FAKE_COLOR,
        label=(
            "Average Precision = "
            f"{average_precision:.3f}"
        ),
    )

    axis.set_title(
        f"Test Precision-Recall Curve - {level_name}",
        fontsize=14,
        fontweight="bold",
        pad=12,
    )

    axis.set_xlabel(
        "Recall",
        fontsize=11
    )

    axis.set_ylabel(
        "Precision",
        fontsize=11
    )

    axis.legend(
        frameon=True,
        loc="lower left"
    )

    axis.grid(
        True,
        alpha=0.25
    )

    register_figure(
        figure,
        filename
    )


# =========================================================
# 7. Frame ve video confusion matrix grafikleri
# =========================================================

create_confusion_matrix_plot(
    frame_predictions,
    "Frame Level",
    "test_confusion_matrix_frame_level.png",
)

create_confusion_matrix_plot(
    video_predictions,
    "Video Level",
    "test_confusion_matrix_video_level.png",
)


# =========================================================
# 8. Frame ve video ROC grafikleri
# =========================================================

create_roc_plot(
    frame_predictions,
    "Frame Level",
    "test_roc_curve_frame_level.png",
)

create_roc_plot(
    video_predictions,
    "Video Level",
    "test_roc_curve_video_level.png",
)


# =========================================================
# 9. Frame ve video precision-recall grafikleri
# =========================================================

create_precision_recall_plot(
    frame_predictions,
    "Frame Level",
    "test_precision_recall_curve_frame_level.png",
)

create_precision_recall_plot(
    video_predictions,
    "Video Level",
    "test_precision_recall_curve_video_level.png",
)


# =========================================================
# 10. Frame-level decision score dağılımı
# =========================================================

figure, axis = plt.subplots(
    figsize=(10, 6),
    dpi=int(CONFIG["figure_dpi"])
)

real_scores = frame_predictions.loc[
    frame_predictions["y_true"] == 0,
    "decision_score",
]

fake_scores = frame_predictions.loc[
    frame_predictions["y_true"] == 1,
    "decision_score",
]

axis.hist(
    real_scores,
    bins=30,
    alpha=0.65,
    label="Real",
    color=REAL_COLOR,
)

axis.hist(
    fake_scores,
    bins=30,
    alpha=0.65,
    label="Fake",
    color=FAKE_COLOR,
)

axis.axvline(
    selected_threshold,
    linestyle="--",
    linewidth=2,
    color="#C53030",
    label=(
        f"Threshold = "
        f"{selected_threshold:.3f}"
    ),
)

axis.set_title(
    "Test Decision Score Distribution - Frame Level",
    fontsize=14,
    fontweight="bold",
    pad=12,
)

axis.set_xlabel(
    "SVM Decision Score",
    fontsize=11
)

axis.set_ylabel(
    "Number of Frames",
    fontsize=11
)

axis.legend(
    frameon=True
)

axis.grid(
    True,
    alpha=0.25
)

register_figure(
    figure,
    "test_decision_score_distribution.png"
)


# =========================================================
# 11. Grafik çözünürlük kalite kontrolü
# =========================================================

figure_audit = pd.DataFrame(
    figure_records
)

assert len(figure_audit) == 9, (
    f"Beklenen grafik sayısı 9, "
    f"oluşturulan: {len(figure_audit)}"
)

assert figure_audit[
    "quality_pass"
].all(), (
    "En az bir grafik minimum "
    "çözünürlük standardını karşılamıyor."
)


write_csv_atomic(
    figure_audit,
    DIRS["metrics"]
    / "figure_quality_audit.csv"
)


print(figure_audit.to_string(
    index=False
))

print(
    "\nHigh-resolution English figures: PASS"
)
```
